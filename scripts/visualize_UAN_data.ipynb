```python
import numpy as np 
import os
import matplotlib.pyplot as plt
```

```python
num_plots = 1
fig, axs = plt.subplots(num_plots, 1, figsize=(12, 3 * num_plots), sharex=True)

for i in range(0, num_plots):
    # Load data
    data = np.load(f'wave_{i+1}.npy', allow_pickle=True).item()
    position_targets = data['position_targets']
    actual_positions = data['actual_positions']
    
    # Compute actual velocities
    actual_velocities = np.diff(actual_positions.flatten(), prepend=actual_positions[0]) / 0.005
    actual_velocities = data['actual_velocities']
    
    # Plot position targets, actual positions, and actual velocities
    axs[i].plot(position_targets, 'k--', label='Position Targets')
    axs[i].plot(actual_positions[:, 3], 'r-', label='Actual Positions')
    
    # axs[i].plot(actual_velocities_sim, 'y-', label='Actual Velocities Sim')
    axs[i].plot(actual_velocities[:, 3], 'b-', label='Actual Velocities')
    axs[i].set_ylabel('Position / Velocity')
    axs[i].legend()

axs[-1].set_xlabel('Time Step')
plt.suptitle('Position Targets, Actual Positions, and Actual Velocities for All Joints')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
num_plots = 10
fig, axs = plt.subplots(num_plots, 1, figsize=(12, 3 * num_plots), sharex=True)

for i in range(0, num_plots):
    # Load data
    try:
        data = np.load(f'left_knee_pyopt_2/npy/gauss_wave_{i+1}.npy', allow_pickle=True).item()
    except:
        continue
    position_targets = data['position_targets']
    actual_positions = data['actual_positions']
    actual_velocities = data['actual_velocities']
    
    # Compute actual velocities
    actual_velocities_from_pos = np.diff(actual_positions, prepend=actual_positions[0]) / 0.01
    
    # Plot position targets, actual positions, and actual velocities
    axs[i].plot(position_targets, 'k--', label='Position Targets')
    axs[i].plot(actual_positions, 'r-', label='Actual Positions')
    
    axs[i].plot(actual_velocities_from_pos/25, 'y-', label='Actual Velocities from pos')
    axs[i].plot(actual_velocities/25, 'b-', label='Actual Velocities')
    axs[i].set_ylabel('Position / Velocity')
    axs[i].legend()

axs[-1].set_xlabel('Time Step')
plt.suptitle('Position Targets, Actual Positions, and Actual Velocities for All Joints')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
# Load validation data
validation_data_path = os.path.join("UAN_data/validation", f"validation_left_hip_roll.npy")
validation_data = np.load(validation_data_path)

# Extract reference positions, real positions, and position history
ref_pos = validation_data[:, 0, :]
real_pos = validation_data[:, 1, :]
pos_hist = validation_data[:, 2, :]

# Plot the data
num_plots = validation_data.shape[0]
fig, axs = plt.subplots(num_plots, 1, figsize=(12, 6 * num_plots), sharex=True)

for i in range(num_plots):
    axs[i].plot(ref_pos[i], 'k--', label='Reference Positions')
    axs[i].plot(real_pos[i], 'r-', label='Real Positions')
    axs[i].plot(pos_hist[i], 'b-', label='Position History')
    axs[i].set_title(f'Validation Data {i+1}')
    axs[i].set_xlabel('Time Step')
    axs[i].set_ylabel('Position')
    axs[i].legend()

plt.tight_layout()
plt.show()
```
